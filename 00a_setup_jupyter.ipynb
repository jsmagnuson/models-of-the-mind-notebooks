## 1. Verify that you have Python

More than likely, you already have Python on your computer. Your first step is to verify
this.

### Macintosh or Linux

Open a terminal (on a Mac, go to Applications :: Utilities and open Terminal.app). Enter
this command:

```bash
python --version
```

You should see a message like this, though your exact version may be different:

```
Python 3.14.3
```

As long as your version number is 3.4 or greater, you can skip to section 3, *Check if
you have conda installed*. If you get a lower version, see the link in section 2 to
install Python.

If instead you get a message like `command not found: python`, first try this:

```bash
python3 --version
```

If you still get the command not found message, you need to install Python.

### Windows

Open a Command Prompt or PowerShell: click on the Start Menu. In the search bar, type
"Command Prompt" and hit Enter. This will open a new Command Prompt window (enter
'PowerShell' instead if you prefer).

Enter this command:

```bash
python --version
```

You should see a message like this, though your exact version may be different:

```
Python 3.10.9
```

That's great, and you can go to section 3, *Check if you have conda installed*.

If instead you get a message like `command not found: python`, first try this:

```bash
python3 --version
```

If you still get a message like `command not found: python3`, or if your version is less
than 3.4, you need to install Python.

## 3. Check if you have conda installed

We will use conda to create a virtual environment where we will install the packages we
need for Jupyter notebooks. You could install directly in your base environment using
pip, but this is not recommended. To see if conda is installed, at a command line
prompt, enter:

```bash
conda --version
```

If you get a message like `conda 24.11.2`, you can move to the next step. If instead you
get a message like `conda not found`, you need to install conda. To do so, go to this
URL and follow the instructions for your operating system:

[https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html](https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html)

---

## 4. Installing the packages we need in a virtual environment using conda

First, we will create a new conda environment called `jupyter-env`.

```bash
conda create --name jupyter-env python=3.10.9 -c conda-forge
conda activate jupyter-env
```

Now that `jupyter-env` is activated, we can install the packages we need. First, tell
conda to use the faster "libmamba" solver, which also prevents some potential problems
with the default solver:

```bash
conda config --set solver libmamba
```

Now install the packages:

```bash
conda install -c conda-forge anywidget ipython ipywidgets "mesa<3" networkx numpy plotly powerlaw matplotlib pandas scipy seaborn scikit-learn statsmodels tk tqdm
```

Note that you will be presented with a much longer list of packages that will be
installed. That is because that list includes packages that the ones we want to install
depend upon. So answer 'y' when asked if you want to proceed.

<div class="alert alert-block alert-warning">

**Let it finish.** Let the command run all the way to the end. Downloading the packages
is *not* the last step: once the download bars finish, conda still has to link the files
into the environment. The installation is finished only when you see `Executing
transaction: done` and your command prompt returns. Do not close the terminal before
that happens.

</div>

<div class="alert alert-block alert-info">

**About** `"mesa<3"`**.** The quotation marks around `"mesa<3"` are required, and so is
the version limit (the `<3` part). Our agent-based modeling notebooks use the Mesa 2
interface; it doesn't work with version 3 (yet).

</div>

Now check that the packages really installed. This is the only reliable way to know.
Copy and paste the whole command below as a single line:

```bash
python -c "import anywidget, IPython, ipywidgets, mesa, networkx, numpy, plotly, powerlaw, matplotlib, pandas, scipy, seaborn, sklearn, statsmodels, tqdm; print('SUCCESS: all packages installed')"
```

<div class="alert alert-block alert-success">

**What you should see.** If you see `SUCCESS: all packages installed`, you are done with
this step. If instead you see an error like `ModuleNotFoundError: No module named
'seaborn'`, then that package did not install. Install that one package by itself and
run the check again, repeating until you get the success message:

```bash
conda install -c conda-forge seaborn
```

</div>

If you ever want to start over with a clean environment, leave the environment, delete
it, and go back to the beginning of step 4:

```bash
conda deactivate
conda env remove --name jupyter-env
```

If you want to use PyTorch (packages for deep learning, not required for the course
notebooks), also do this:

```bash
conda install pytorch torchvision -c pytorch
```

## 5. Install and test Jupyter lab

Now we install jupyterlab.

```bash
conda install -c conda-forge jupyterlab
conda install -c conda-forge ipympl
```

Now try to open Jupyter lab with the following command; it should open a new window in
your default web browser (Safari, Chrome, Edge, etc.):

```bash
jupyter lab
```

<div class="alert alert-block alert-info">

**If something goes wrong.** Of course, if you get errors at this stage or any other,
ask me for help. Try to take a screenshot of the error, or even better, copy and paste
the actual text of the error message into your email if possible. Otherwise, the
installation steps are done at this point!

</div>

Subsequently, whenever you want to run Jupyter lab with this environment, you need to
first activate the environment, and then call `jupyter lab`. You can then navigate to
the directory where you have the course notebooks, or you can call Jupyter lab from that
directory (much easier); I'll show you this in class:

```bash
conda activate jupyter-env
jupyter lab
```

If you ever forget what you called your environment, you can use this command to list
all available environments:

```bash
conda env list
```

## 7. Open course notebooks in Jupyter lab

At the command line (Terminal for Mac or Linux, Command Prompt or PowerShell for
Windows), navigate to the directory/folder where you put the course notebooks. If you
don't know how to do that, just skip that step for now. At the command line, enter this
command (with a space this time):

```bash
jupyter lab
```

If you executed this command from the folder where the notebooks are located, you will
see the notebook files listed in the Jupyter lab file browser on the left. If you could
not navigate to the directory location from the command line, just use the Jupyter lab
interface to navigate there. However, you will want to learn how to navigate to that
location — so please ask if you do not know how.
