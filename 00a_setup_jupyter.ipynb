
# Setting Up for Working With Jupyter Notebooks

*Last update: 09/04/2026, Jim Magnuson*

It is crucial to set up your computer for running **JupyterLab** during the first week of classes.
Some notebooks require libraries that **are not available on Google Colab**, so you must install Python and Jupyter locally.

If you do not have a computer you can use for this course, please talk to the instructor ASAP.

If you are already comfortable with Python environments (e.g., conda), you may use your own setup. Below are the simplest recommended instructions.

---

## 1. Verify That You Have Python

### macOS or Linux

Open **Terminal** (macOS: Applications → Utilities → Terminal.app) and run:

```bash
python --version
```

You should see something like:

```
Python 3.14.3
```

If your version is **3.4 or higher**, skip ahead to section 3.
If you get "command not found: python", try `python3 --version`. If that also fails, install Python (see section 2).

---

### Windows

Open **Command Prompt** or **PowerShell** (Start Menu → search "Command Prompt" or "PowerShell") and run:

```bash
python --version
```

You should see something like:

```
Python 3.10.9
```

If your version is **3.4 or higher**, continue to section 3.
If you get "command not found: python", try `python3 --version`. If that also fails, or your version is older than 3.4, install Python.

---

## 2. Install Python (Only if Needed)

Download and install Python from:
[https://www.python.org/downloads/](https://www.python.org/downloads/)

## 3. Check if You Have conda Installed

We will use **conda** to create a virtual environment for the course. You could install directly into your base environment using pip, but this is not recommended.

At your command prompt, run:

```bash
conda --version
```

If you see something like:

```
conda 24.11.2
```

you're good.
If instead you see "conda not found", install conda following the instructions for your OS:

[https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html](https://docs.conda.io/projects/conda/en/stable/user-guide/install/index.html)

---

## 4. Create a Virtual Environment and Install Required Packages

### Create the environment

```bash
conda create --name jupyter-env python=3.10.9 -c conda-forge
conda activate jupyter-env
```

### Tell conda to use the libmamba solver

Now that `jupyter-env` is activated, first tell conda to use the faster "libmamba" solver, which also prevents some potential problems with the default solver:

```bash
conda config --set solver libmamba
```

### Install required packages

```bash
conda install -c conda-forge anywidget ipython ipywidgets "mesa<3" networkx numpy plotly powerlaw matplotlib pandas scipy seaborn scikit-learn statsmodels tk tqdm
```

You will be presented with a much longer list of packages than the ones above. That is because the list includes packages that the ones we want depend upon. Answer **y** when asked if you want to proceed.

<div class="alert alert-block alert-warning">

**Let it finish.** *Downloading the packages is not the last step: once the download bars finish, conda still has to link the files into the environment. The installation is complete only when you see* `Executing transaction: done` *and your command prompt returns. Do not close the terminal before that happens.*

</div>

<div class="alert alert-block alert-info">

**About** `"mesa<3"`**.** *The quotation marks are required, and so is the version limit (the* `<3` *part). Our agent-based modeling notebooks use the Mesa 2 interface; it doesn't work with version 3 (yet).*

</div>

### Check that the packages really installed

This is the only reliable way to know. Copy and paste the whole command below as a single line:

```bash
python -c "import anywidget, IPython, ipywidgets, mesa, networkx, numpy, plotly, powerlaw, matplotlib, pandas, scipy, seaborn, sklearn, statsmodels, tqdm; print('SUCCESS: all packages installed')"
```

If you see `SUCCESS: all packages installed`, you are done with this step. If instead you see an error like `ModuleNotFoundError: No module named 'seaborn'`, then that package did not install. Install that one package by itself and run the check again, repeating until you get the success message:

```bash
conda install -c conda-forge seaborn
```

### Starting over with a clean environment

If you ever want to start over, leave the environment, delete it, and go back to the beginning of section 4:

```bash
conda deactivate
conda env remove --name jupyter-env
```

### (Optional) Install PyTorch

Packages for deep learning. These are not required for the course notebooks.

```bash
conda install pytorch torchvision -c pytorch
```

## 5. Install and Test JupyterLab

```bash
conda install -c conda-forge jupyterlab
conda install -c conda-forge ipympl
```

Now try to open JupyterLab. It should open a new window in your default web browser (Safari, Chrome, Edge, etc.):

```bash
jupyter lab
```

If you get errors at this stage or any other, ask me for help. Take a screenshot of the error, or better still, copy and paste the actual text of the error message into your email if possible. Otherwise, the installation steps are done at this point!

### Later use

Each time you want to run JupyterLab, first activate the environment, then launch:

```bash
conda activate jupyter-env
jupyter lab
```

### If you forget your environment name

```bash
conda env list
```

Navigate (in Terminal/Command Prompt) to the directory where you saved the notebooks.

```bash
jupyter lab
```

Otherwise, use the JupyterLab interface to navigate to the folder.
